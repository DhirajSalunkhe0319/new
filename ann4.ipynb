4.With a suitable example demonstrate the perceptron learning law with its decision regions using python. Give
the output in graphical form

```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
X=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]])
T=np.array([0,0,0,1])
```

```python
learning_rate=0.1
epochs=5
w=np.zeros(2)
bias=0
epoch_errors =[]
```

```python
def activation(net):
    if net>=0:
        return 1
    else:
        return 0
```

```python
for epoch in range(epochs):
    total_error=0

    for i in range(len(X)):
        x=X[i]
        net=np.dot(w,x)+bias
        y=activation(net)

        current_step_error=T[i]-y

        w=w+(learning_rate*current_step_error*x)
        bias=bias+(learning_rate*current_step_error)

        total_error=total_error+abs(current_step_error)

    epoch_errors.append(total_error)
```

```python
print("Training completed in ",epoch+1 ,"epochs")
print("\nFinal weights =",w)
print("\nFinal bias =",bias)
print("\nOutput verification ")

for i in range(len(X)):
    net=np.dot(w,X[i])+bias
    y=activation(net)

    print("Input ",X[i] ,"Target ",T[i],"predicted ",y)
```

```python
print(epoch_errors)
```

```python
#convergence graph of perceptron learning low 
plt.figure(figsize=(10,5))


plt.plot(range(1 , len(epoch_errors)+1),epoch_errors,marker="o",color='blue',linewidth=2)
plt.title("Convergence of perceptron learning low")
plt.xlabel("Epoch")
plt.ylabel("Total misclassification error")
plt.grid(True)
plt.show()
```

```python
plt.scatter(X[T==0,0],X[T==0,1],color='red',marker='s',s=100,label="class 0 ")
plt.scatter(X[T==1,0],X[T==1,1],color='blue',marker='o',s=100,label="class 1")


x=np.linspace(-0.5,1.5,100)
y=1.5-x

xx,yy=np.meshgrid(
    np.linspace(-0.5,1.5,200),
    np.linspace(-0.5,1.5,200))
Z=xx+yy-1.5
plt.contourf(xx,yy,Z>=0,alpha=0.3,
             cmap="plasma")

plt.plot(x,y,color="black",linewidth=2,label="Desicion boundary")

plt.title("Desicion regions -perceptron")
plt.xlabel("Inputx1")
plt.ylabel("Inputx2")
plt.xlim(-0.5,1.5)
plt.ylim(-0.5,1.5)
plt.grid(True)
plt.legend()
plt.show()
```
