Solve a simple grid world problem using reinforcement earning, where an agent learns to reach a target location while avoiding obstacles. How to delive this problem statements to students


```python
import numpy as np
import random
```

```python
# Grid size
grid_size = 4

# Start position
start = (0, 0)

# Goal position
goal = (3, 3)

# Obstacles
obstacles = [(0, 2), (1, 2), (2, 1)]
```

```python
# ------------------------------------------------
# 2. Actions
# ------------------------------------------------

actions = {
    0: (-1, 0),   # Up
    1: (1, 0),    # Down
    2: (0, -1),   # Left
    3: (0, 1)     # Right
}

# ------------------------------------------------
# 3. Rewards
# ------------------------------------------------

goal_reward = 10
obstacle_reward = -10
step_reward = -1
outside_reward = -5
```

```python
# ------------------------------------------------
# 4. Q-table
# ------------------------------------------------

Q = np.zeros((grid_size, grid_size, 4))

# ------------------------------------------------
# 5. Parameters
# ------------------------------------------------

learning_rate = 0.8
discount_factor = 0.9

episodes = 1000

epsilon = 1.0
epsilon_decay = 0.995
epsilon_min = 0.01
```

```python
# ------------------------------------------------
# 6. Environment
# ------------------------------------------------

def take_action(state, action):

    row, col = state

    move = actions[action]

    new_row = row + move[0]
    new_col = col + move[1]

    # Outside grid
    if new_row < 0 or new_row >= grid_size or \
       new_col < 0 or new_col >= grid_size:

        return state, outside_reward, False

    new_state = (new_row, new_col)

    # Obstacle
    if new_state in obstacles:
        return state, obstacle_reward, False

    # Goal
    if new_state == goal:
        return new_state, goal_reward, True

    # Normal cell
    return new_state, step_reward, False
```

```python
# 7. Epsilon-greedy action selection
# ------------------------------------------------

def choose_action(state, epsilon):

    if random.random() < epsilon:

        # Exploration
        return random.randint(0, 3)

    else:

        # Exploitation
        row, col = state

        return np.argmax(Q[row, col])

```

```python
# ------------------------------------------------
# 8. Training
# ------------------------------------------------

for episode in range(episodes):

    state = start

    for step in range(100):

        # Select action
        action = choose_action(state, epsilon)

        # Perform action
        next_state, reward, done = take_action(
            state, action
        )

        row, col = state

        next_row, next_col = next_state

        # Current Q-value
        current_q = Q[row, col, action]

        # Maximum future Q-value
        max_future_q = np.max(
            Q[next_row, next_col]
        )

        # Q-learning formula
        new_q = current_q + learning_rate * (
            reward
            + discount_factor * max_future_q
            - current_q
        )

        # Update Q-table
        Q[row, col, action] = new_q

        # Move to next state
        state = next_state

        # Stop when goal is reached
        if done:
            break

    # Reduce exploration
    epsilon = max(
        epsilon_min,
        epsilon * epsilon_decay
    )


# ------------------------------------------------
# 9. Find best path
# ------------------------------------------------

def get_best_path():

    state = start

    path = [state]

    for step in range(20):

        if state == goal:
            break

        row, col = state

        # Select best action
        action = np.argmax(Q[row, col])

        next_state, reward, done = take_action(
            state, action
        )

        if next_state == state:
            break

        state = next_state

        path.append(state)

        if done:
            break

    return path


# ------------------------------------------------
# 10. Display path
# ------------------------------------------------

def display_grid(path):

    for row in range(grid_size):

        for col in range(grid_size):

            position = (row, col)

            if position == start:
                symbol = "S"

            elif position == goal:
                symbol = "G"

            elif position in obstacles:
                symbol = "X"

            elif position in path:
                symbol = "*"

            else:
                symbol = "."

            print(symbol, end=" ")

        print()


# ------------------------------------------------
# 11. Output
# ------------------------------------------------

path = get_best_path()

print("\nLearned Path:")
print(path)

print("\nGrid:")
display_grid(path)
```
